### Monte Carlo Error Propagation (4 pt) <font color='red'><b>COURSEWORK</b></font>

One of the classical Physic Lab experiment is to use pendulum to calculate the local gravitational acceleration, ``g``, from the measurements of the pendulum length ``L`` and the oscillation period ``T``.

The formula is expressed as follows:

$$ g=\frac{4\pi^2 L}{T^2} $$

You have got a set of measurements of $L=1.00\pm0.01\ \mathrm{(m)}$ and $T=2.01\pm0.02\ \mathrm{(s)}$.

* Like in Physics Lab, use Tyler expansion described above to propagate the error. Write your derivation in Markdown.

* Calculate the error of the derived ``g``.

* Apply Monte Carlo error propagation to estimate the value and uncertainty of ``g``.

* Make appropriate comments.

* Compare the two results.

Error: 
$2\frac{0.02}{2.01}+\frac{0.01}{1}=0.03 (fractional error)$

Monte Carlo:
$\frac{\partial g}{\partial L} = \frac{4\pi^2}{T^2} = \frac{4\pi^2}{(2.01)^2} = 9.7716 \newline$ 
$\frac{\partial g}{\partial T} = -\frac{8\pi^2 L}{T^3} = -\frac{8\pi^2 (1.0)}{(2.01)^3} =-9.723$

Error propagation: 
$\sigma_g^2 =  \left( \frac{\partial g}{\partial L} \sigma_L \right)^2 + \left( \frac{\partial g}{\partial T} \sigma_T \right)^2 $

$\sigma_g = \sqrt{ (9.7716 \times 0.01)^2 + (-9.723 \times 0.02)^2 }=0.22 (rounded = 0.2)$ 

$g = \frac{4\pi^2 (1.0)}{(2.01)^2} = 9.772 \text{m/s}^2$

error = $9.772 \times 0.03=0.29 \text{(rounded = 0.3)}$

The error propagation formula gives a lower error than the error calculations from the basic formula with rational errors. This may help in making observations more accurate, including the uncertainties.

### Plotting Hertzsprung–Russell (HR) diagram (6 pt) <font color='red'><b>COURSEWORK</b></font>

``Hertzsprung-Russell (HR) diagram``, which indicates the stellar evolution, is an important astrophysical plot graphing stellar luminosity (brightness, absolute magnitude) against surface temperature (or colour). This diagram clearly states the evolutionary pathways of stars and where they are in their lifetime. With this, we classify several important stages of a star's life -- main sequence stars, giants, or white dwarfs. 

In this exercise, a combined APOGEE/Gaia star catalog has been prepared for you to create an HR diagram -- download the catalogue from the BrightSpace: ``APOGEE+Gaia_stars.fits``

**<span style="color:red">Warning</span>**: note that from observation, we only have apparent magnitude (i.e. observed magnitude). To convert into absolute magnitude requires **Distance Modulus**: 

$$ M_G = m_G + 5\log10{(\mathrm{parallax}/1000)} + 5 $$

* Apply a rough binary classification scheme to separate giant branch and main sequence (MS) stars.

````python
from astropy.io import fits
from astropy.table import Table
import numpy as np
# Open the FITS file
hdul = fits.open('APOGEE_stars.fits')

# Access the primary data (in this case, it is a Table)
data = hdul[1].data  # Often, the first extension contains the table
colnames = hdul[1].columns.names # List of the column names
#error_ratio is the ratio in the fourth point for the data selection criteria
error_ratio = data['GAIAEDR3_PARALLAX']/data['GAIAEDR3_PARALLAX_ERROR']
quality_data = data[(data['STARFLAG']==0)&(data['ASPCAPFLAG']==0)\
                   & (data['SNR']>10)&(error_ratio>5)&\
                    (~np.isnan(data['GAIAEDR3_PHOT_BP_MEAN_MAG'])) & \
                    (~np.isnan(data['GAIAEDR3_PHOT_RP_MEAN_MAG'])) & \
                    (~np.isnan(data['GAIAEDR3_PHOT_G_MEAN_MAG']))]
#grabs the table and shuffles the data

table_randomized = quality_data[np.random.permutation(len(quality_data))]

#slice the first 80%
quantity = int(0.8*len(quality_data))
to_plot = table_randomized[:quantity]

colour = to_plot['GAIAEDR3_PHOT_BP_MEAN_MAG']-to_plot['GAIAEDR3_PHOT_RP_MEAN_MAG']
app_mag = to_plot['GAIAEDR3_PHOT_G_MEAN_MAG'] ## not sure if this is the correct value for apparent magnetude
abs_mag = app_mag +5*np.log10(to_plot['GAIAEDR3_PARALLAX']/1000)+5

GB_bool = (abs_mag<3.5)&(colour>0.8) #condition
GB = to_plot[GB_bool] #where condition is true
wd_bool = (colour <= 0.8) & (abs_mag > 10)
WD = to_plot[wd_bool]
ms_bool = ~GB_bool & ~wd_bool
MS = to_plot[ms_bool]

print(colnames)
print(GB[:2])

import matplotlib.pyplot as plt
plt.figure(figsize=(8, 6))
plt.scatter(colour[ms_bool], abs_mag[ms_bool], s=1, c='blue', label='MS', alpha=0.5)
plt.scatter(colour[GB_bool], abs_mag[GB_bool], s=1, c='red', label='Giant Branch')

plt.gca().add_patch(plt.Rectangle((0, 10), 0.8, 5, color='gray', alpha=0.3, label='White Dwarf Region'))

plt.xlabel("BP - RP")
plt.ylabel(r"$M_G$")
plt.xlim(-0.5,4.5)
plt.ylim(-2.5,15)

plt.annotate('Giant Branch', 
             xy=(1, 0.5),         
             xytext=(2.2, -1),    
             color='black', fontsize=12, fontweight='bold',
             arrowprops=dict(color='black', shrink=0.05, width=1, headwidth=5))
plt.annotate('MS', 
             xy=(1.1, 4.5),       # Pointing at the dense blue cluster
             xytext=(2.0, 3.5),   # Where the 'MS' text sits
             color='black', fontsize=12, fontweight='bold',
             arrowprops=dict(color='black', shrink=0.05, width=1, headwidth=5))
##i chose black for the colors because they wouldn't have been visible in that


plt.gca().invert_yaxis()
plt.legend()
plt.show()
````
