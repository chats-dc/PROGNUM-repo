```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm as normal

norm = np.random.normal(loc=185,scale=8,size=100000)


plt.hist(norm,range=[min(norm),max(norm)],bins=40,label="people")


mean = np.mean(norm)
deviation =np.std(norm)
plt.axvline(mean,color='red',label=f'mean={mean:.2f} \n Standard deviation {deviation}',linewidth=0.5)

#plt.fill_betweenx(norm,mean,mean+deviation,color="orange",alpha=1)
plt.axvspan(mean-deviation,mean+deviation,color="blue",alpha=0.5,label=f'standard deviation \n {deviation:.2f}') #used this video: https://www.youtube.com/watch?v=NfjMtTM9Ip0
plt.legend()
plt.show()

h=185 #can also be a value inputted
cdf = normal.cdf(h,scale=deviation,loc=mean)

#within 1 standard deviation:
cdfmin = normal.cdf(mean-deviation,scale=deviation,loc=mean)
cdfmax = normal.cdf(mean+deviation,scale=deviation,loc=mean)
cdfrange = cdfmax-cdfmin

print(f"probability of people shorter than {h} is: {cdf*100:.2f} percent")
print(f"Probability of people within a deviation of {h} is: {cdfrange*100:.2f}")
```

```python

def funcfilter(func,seq):
    filtered_list=[]
    
    for i in range(len(seq)):
        if seq[i]>0:
            filtered_list.append(seq[i])
    loglist = func(filtered_list)
    return loglist

sq = np.array([-100, -10, 0, 10, 100])
result = funcfilter(np.log10, sq)
print("Result log10:", result)
```

```python
from astropy.cosmology import Planck18
from scipy import integrate
import astropy.units as u
import numpy as np

H0 = Planck18.H0.value 
Om0 = Planck18.Om0
Ode0 = Planck18.Ode0


#im using x instead of z' because i don't know how jupyter would respond to '

def H(x):
    H = H0*np.sqrt(Om0*((1+x)**3)+Ode0)
    return H
def F(x): #function to be integrated
    y = 1/((1+x)*H(x))
    return y

a=0
b=2
integral = integrate.quad(F, a, b) #gotta convert from mpc/km s to gigayears
time = integral[0]*3.086*10**19/(31556926)/1000000000 #31556926 seconds in a year 

print(time)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm as normal
from scipy.stats import multivariate_normal

x = np.random.normal(loc=0,scale=1,size=100000)
y = np.random.normal(loc=2,scale=1,size=100000)

xplot = plt.hist(x,bins=30,density=True,label="x",color='mediumblue',alpha=0.5,edgecolor='black')
yplot = plt.hist(y,bins=30,density=True,label="y",color='red',alpha=0.5,edgecolor='black')

x1=np.linspace(x.min(),x.max(),100000,dtype=np.dtype(np.float32))
y1=np.linspace(y.min(),y.max(),100000,dtype=np.dtype(np.float32))

plt.plot(x1,normal.pdf(x1,0,1),label = 'X(PDF)',linestyle=":",color='black',linewidth=4)
plt.plot(y1,normal.pdf(y1,2,1),label = 'Y(PDF)',linestyle="--",color='black',linewidth=4)

plt.annotate('$\mathcal{N}(0,1)$',(-2,0.15),(-4.5,0.15),color='darkblue'\
             ,arrowprops=dict(arrowstyle='-|>', color ='darkblue',lw=5))
plt.annotate("$\mathcal{N}(2,1)$",(4,0.15),(5,0.15),color='red'\
             ,arrowprops=dict(arrowstyle='-|>', color ='red',lw=5))

plt.xticks(np.arange(-6,9,1))
plt.xlabel('Values')
plt.title('1D Histogram + PDF')
plt.grid(axis = 'y',linestyle = '--')
plt.legend()
plt.show()

count,xedges,yedges,image = plt.hist2d(x, y, bins=(30, 30), cmap=plt.cm.Blues)
count_flat = count.flatten()
sorted_ = np.sort(count_flat)[::-1] #descending order
cum_sum=np.cumsum(sorted_)
cum_sum /=cum_sum[-1] #divides every value by the last value
print(cum_sum)
levelvals = [0.68,0.95,0.997]
levels=[]

for i in levelvals:
    index = np.argmin(np.abs(cum_sum-i)) #finds the index closest to the level
    levels.append(sorted_[index])
    
levels = sorted(list(set(levels))) 

fig = plt.hist2d(x, y, bins=30, cmap='Blues')
plt.colorbar()

# centers of the bins
x_centers = (xedges[:-1] + xedges[1:]) / 2
y_centers = (yedges[:-1] + yedges[1:]) / 2

contour = plt.contour(x_centers, y_centers, count.T, levels=levels, colors='red', linewidths=2)
labels = {levels[0]: r'$3\sigma$', 
          levels[1]: r'$2\sigma$', 
          levels[2]: r'$1\sigma$'}

plt.clabel(contour, inline=True, fontsize=10, fmt=labels)

plt.xticks(np.arange(-5,5,1))
plt.xlabel('X')
plt.ylabel("Y")
plt.title('2D Histogram + Contour')
plt.grid(linestyle = '--',color="k")
plt.show()

```
