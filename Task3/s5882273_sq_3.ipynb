```python
from scipy.constants import G

masses = [1.9891e+30, 1.8986e+27, 5.6846e+26, 
          1.0243e+26, 8.6810e+25, 5.9736e+24, 
          4.8685e+24, 6.4185e+23, 3.3022e+23, 
          7.349e+22, 1.25e22]
radii  = [696340, 71492, 60268, 24764, 
          25559, 6378.1, 6051.8, 3396.2, 
          2439.7, 1738.1, 1195]
names  = ['Sun', 'Jupiter', 'Saturn', 'Neptune', 
          'Uranus', 'Earth', 'Venus', 'Mars', 
          'Mercury', 'Moon', 'Pluto']


vesc=[]
dictvels ={}
for i in range(len(masses)):
    
    v = (2*G*masses[i]/radii[i])**0.5
    vesc.append(v)
    print(f"{masses[i]:.2e}")
    dictvels[names[i]]=v
    
print(dictvels)

```

### 3.10

```python
def cylinder(radius,height):
    area = 2*pi*radius**2+2*pi*radius*height
    volume = pi*radius**2*height
    return area,volume

rval = input("enter a radius:")
hval = input("enter a height:")
r = float(rval)
h = float(hval)

area,volume=cylinder(r,h)
ratio = area/volume
print(f"Area is {area:.2f} and volume is {volume:.2f}")
print(f"Ratio is {ratio:.2f}")
```

### 3.17

```python
import numpy as np

def stdev(array):
    deviat = np.sqrt(np.sum((array-np.mean(array))**2)/len(array))
    return deviat
    
    
a = input('Enter an array:')
arraya = np.fromstring(a, dtype=float, sep=',')
deviation = np.std(arraya)
deviat=stdev(arraya)

print(f'Using np.std: {deviation}')
print("Using function:", deviat)
```

```python
# try this np.fromstring('1 2', dtype=int, sep=' ')
from io import StringIO
# stringio acts like a file
#triple quotes below makes the whole thing part of the string
a=StringIO('''18.49 2094.75 
35.48 2320.87
35.48 1224.9
11.34 872.13
7.12  432.2
9.82  751
23.66 1410
44    2700
66.7  5292
52.4  2550
55    5253
25.6  2419
26.19 2158''')
b=np.loadtxt(a)
dist=b[:,0] #this is how to slice columns
#[row,column].... colon means all in
# example: trial = b[0,:] gives elements in first row
vel = b[:,1]


##x is dist, y is vel 
x_bar = sum(dist)/len(dist) ##mean/average
y_bar=sum(vel)/len(vel)

prod_num=0 #numerator of b fraction
prod_denom=0
for i in range(len(b)):
    prod_num=prod_num+(dist[i]-x_bar)*(vel[i]-y_bar)
    prod_denom=prod_denom+(dist[i]-x_bar)**2
    #this works but try to use numpy functions

b=prod_num/prod_denom
aval=y_bar-(b*x_bar)
print(f'b={b:.2f}')
print(f'a={aval:.2f}')
```

```python
from matplotlib import pyplot as plt
import numpy as np
x = []
y=[]
a=-5
for a in range(-5,6):
    x.append(a)
    b=((np.exp(a)+np.exp(-a))/2)
    y.append(b)

plt.plot(x,y,marker=".",color="black",label="$\cosh(x)$")
plt.title("cosh function",color="red")
plt.xlabel("x value", fontsize=14, color="grey")
plt.ylabel("function", fontsize=14, color="grey")
plt.grid()
plt.xlim(-6,6)
plt.legend()
plt.show()

h = np.arange(x[0],max(x)+1)

print(h)
```
